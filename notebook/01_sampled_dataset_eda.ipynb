```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Importing required functions from data_loader.py

```python
import sys 
sys.path.append("..")
from src.data_loader import eda_stats
```

Loading the sampled dataset (n=1000)

```python
filepath = "../results/data/sampled_n_200.csv"
sampled_df = pd.read_csv(filepath)
```

```python
txc_dist, co_occ_txc_grpd, idnt_dist, co_occ_idnt_grpd, cmnt_summary = eda_stats(sampled_df)
```

```python
graph_color = "#EF3E36"
text_color = "#96031A"
sns.set_style({
    "axes.facecolor": "#FFE8D4",
    "figure.facecolor": "#FFE8D4",

    "grid.color": graph_color,
    "axes.edgecolor": graph_color,
    "axes.grid": True,
    "axes.axisbelow": True,
    
    "axes.labelcolor": text_color,
    "text.color": text_color,
    "font.family": "arial",
    "xtick.color": text_color,
    "ytick.color": text_color,

    "xtick.bottom": False,
    "xtick.top": False,
    "ytick.left": False,
    "ytick.right": False,

    "axes.spines.left": False,
    "axes.spines.bottom": True,
    "axes.spines.right": False,
    "axes.spines.top": False,
})
```

H bar chart of toxicity subtypes distribution

```python
txc_dist = dict(sorted(txc_dist.items(), key=lambda item: item[1], reverse=True))
#fig = plt.figure(figsize=(10, 7))
sns.barplot(
    x = list(txc_dist.values()),
    y = list(txc_dist.keys()),
    color = graph_color,
    edgecolor = "none",
    height = 0.65
)

plt.title("Number of Comments per Toxicity Subtype (Non-distinct)")
plt.tight_layout()
plt.show()
```

V bar chart of co-occurence of subtoxicity types in comments

```python
sns.barplot(
    x = co_occ_txc_grpd.index.tolist(),
    y = co_occ_txc_grpd.values.tolist(),
    color = graph_color,
    edgecolor = "none",
    width = 0.65
)

plt.title("Co-occurence of Toxicity Subtypes in Comments")
plt.tight_layout()
plt.show()
```

H barchart of identity tags distribution

```python
#fig = plt.figure(figsize=(10, 7))
sns.barplot(
    x = list(idnt_dist.values()),
    y = list(idnt_dist.keys()),
    color = graph_color,
    edgecolor = "none",
    height = 0.65
)

plt.title("Number of Comments per Identity Tag (Non-distinct)")
plt.tight_layout()
plt.show()
```

V bar chart of co-occurence of identity tags in comments

```python
sns.barplot(
    x = co_occ_idnt_grpd.index.tolist(),
    y = co_occ_idnt_grpd.values.tolist(),
    color  = graph_color,
    edgecolor = "none",
    width = 0.65
)

plt.title("Co-occurence of Identity Tags in Comments")
plt.tight_layout()
plt.show()
```
